## **Pregunta 5**

¿Cuál es la potencia bruta y neta de la Central localizada en Salta (Argentina)?

```python
df[df['region_nombre'] == "Salta (Argentina)"][['sistema','subsistema', 'propietario','tipo_de_energia','potencia_bruta_mw', 'potencia_neta_mw']]
```

```python
df.head()
```

Filtrando solo plantas en operación, se tiene:

```python
df1 = df[(df['estado'] == "En Operacion") &
 (df['region_nombre'] != "Salta (Argentina)") &
  (df['sistema'].isin(['SEN', 'SEA', 'SEM']))]
df1.head()
```

El número de observaciones por sistema es:

```python
df1['sistema'].value_counts()
```

```python
#df1['potencia_bruta_mw'] = df1['potencia_bruta_mw'].str.replace(',', '.', regex=False).astype(float)
df1.groupby('region_nombre')[['potencia_bruta_mw', 'potencia_neta_mw']].sum()
```

```python
print(df1.columns)
```

```python
df1.groupby('tipo_de_energia')['potencia_neta_mw'].sum()
```

```python
df1['tipo_de_energia2'] = df1['tipo_de_energia'].replace({
    'Solar Fotovoltaica': 'Solar',
    'Solar-CSP': 'Solar',
    'Carbón - Petcoke': 'Carbón',
    'Petcoke': 'Carbón',
    'Mini Hidráulica Pasada': 'Hidráulica',
    'Hidráulica Pasada': 'Hidráulica',
    'Hidráulica Embalse': 'Hidráulica',
    'Cogeneración': 'Gas Natural',
    'Propano': 'Gas Natural',
    'Biogas': 'Biomasa',
    'Biomasa-Petróleo N°6': 'Biomasa',
    'Fuel Oil Nro. 6': 'Petróleo Diesel'

})

print("Nuevas categorías de 'tipo_de_energia' después de agrupar:")
df1['tipo_de_energia2'].value_counts()
```

```python
df1.groupby('tipo_de_energia2')['potencia_neta_mw'].sum()
```

```python
df1.groupby('sistema')['potencia_neta_mw'].sum().round(0)
```

```python
df1[df1['sistema'] == "SEA"].groupby('tipo_de_energia2')['potencia_neta_mw'].sum().round(0)
```

```python
df1[df1['sistema'] == "SEM"].groupby('tipo_de_energia2')['potencia_neta_mw'].sum().round(0)
```

```python
df1[df1['sistema'] == "SEN"].groupby('tipo_de_energia2')['potencia_neta_mw'].sum().round(0)
```

```python
import plotly.express as px

df_sen = df1[df1['sistema'] == "SEN"]

# Agrupar por tipo_de_energia y sumar la potencia bruta
df_treemap = df_sen.groupby('tipo_de_energia2')['potencia_neta_mw'].sum().reset_index()

# Crear el treemap
fig = px.treemap(df_treemap, path=[px.Constant("SEN"), 'tipo_de_energia2'],
                 values='potencia_neta_mw',
                 title='Distribución de Potencia Neta (MW) por Tipo de Energía en la Región de Los Lagos')

fig.update_layout(margin = dict(t=50, l=25, r=25, b=25))
fig.show()
```
